```python
df_clean = df.drop_duplicates()

cols = ["Operating_Cash_Flow_FY25_INR_Cr", "Receivable_Days", "Payment_Overdue_Days"]

for col in cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())

df_clean.isnull().sum()
```

```python
df_clean["Revenue_Growth_%"] = (
    (df_clean["Revenue_FY25_INR_Cr"] - df_clean["Revenue_FY24_INR_Cr"])
    / df_clean["Revenue_FY24_INR_Cr"]
) * 100

df_clean[["Borrower_ID", "Revenue_Growth_%"]].head(10)
```

```python
df_clean.sort_values("Revenue_Growth_%",
    ascending=True
).head(10)
```

```python
df_clean["EBITDA_Margin_%"] = (
    df_clean["EBITDA_FY25_INR_Cr"]
    / df_clean["Revenue_FY25_INR_Cr"]
) * 100

df_clean[
    ["Borrower_ID", "Industry",
     "Revenue_FY25_INR_Cr",
     "EBITDA_FY25_INR_Cr",
     "EBITDA_Margin_%"]
].head(10)
```

```python
df = df_clean[df_clean["EBITDA_Margin_%"] < 8]
df = df.sort_values("EBITDA_Margin_%")
df[["Borrower_ID", "Industry", "EBITDA_Margin_%"]]
```

```python

df_clean["Debt_to_EBITDA"] = df_clean["Total_Debt_INR_Cr"] / df_clean["EBITDA_FY25_INR_Cr"]

df = df_clean.sort_values("Debt_to_EBITDA", ascending=False).head(10)

df[["Borrower_ID", "Total_Debt_INR_Cr", "EBITDA_FY25_INR_Cr", "Debt_to_EBITDA"]]
```

```python
df_clean["Interest_Coverage"] = df_clean["EBITDA_FY25_INR_Cr"] / df_clean["Interest_Expense_FY25_INR_Cr"]

df = df_clean.sort_values("Interest_Coverage").head(10)

df[["Borrower_ID", "EBITDA_FY25_INR_Cr", "Interest_Expense_FY25_INR_Cr", "Interest_Coverage"]]
```

```python
df_clean["Current_Ratio"] = df_clean["Current_Assets_INR_Cr"] / df_clean["Current_Liabilities_INR_Cr"]

df = df_clean.sort_values("Current_Ratio").head(10)

df[["Borrower_ID", "Current_Assets_INR_Cr", "Current_Liabilities_INR_Cr", "Current_Ratio"]]
```

```python
df_clean["Liquidity_Status"] = df_clean["Current_Ratio"].apply(
    lambda x: "Low" if x < 1 else "Moderate" if x < 1.5 else "Good"
)
df_clean[["Borrower_ID", "Current_Ratio", "Liquidity_Status"]].head(10)
```

```python
df_clean["Cash_Conversion_Cycle"] = df_clean["Receivable_Days"] + df_clean["Inventory_Days"] - df_clean["Payable_Days"]
df = df_clean.sort_values("Cash_Conversion_Cycle", ascending=False).head(10)
df[["Borrower_ID", "Industry", "Receivable_Days", "Inventory_Days", "Payable_Days", "Cash_Conversion_Cycle"]]
```

```python
df = df_clean.groupby("Industry").agg({
    "Borrower_ID": "count",
    "Revenue_Growth_%": "mean",
    "EBITDA_Margin_%": "mean",
    "Debt_to_EBITDA": "mean",
    "Interest_Coverage": "mean",
    "Current_Ratio": "mean"
}).round(2)

df
```

```python
df_clean["Revenue_Decline"] = df_clean["Revenue_Growth_%"] < 0
df_clean["High_Leverage"] = df_clean["Debt_to_EBITDA"] > 4
df_clean["Low_Interest_Coverage"] = df_clean["Interest_Coverage"] < 2
df_clean["Weak_Liquidity"] = df_clean["Current_Ratio"] < 1
df_clean["Negative_Cash_Flow"] = df_clean["Operating_Cash_Flow_FY25_INR_Cr"] < 0
df_clean["Payment_Overdue"] = df_clean["Payment_Overdue_Days"] > 30

df = df_clean[["Borrower_ID", "Revenue_Decline", "High_Leverage", "Low_Interest_Coverage", "Weak_Liquidity", "Negative_Cash_Flow", "Payment_Overdue"]]

df
```

```python
df_clean["Warning_Score"] = df_clean[[
    "Revenue_Decline",
    "High_Leverage",
    "Low_Interest_Coverage",
    "Weak_Liquidity",
    "Negative_Cash_Flow",
    "Payment_Overdue"
]].sum(axis=1)

df = df_clean.sort_values("Warning_Score", ascending=False).head(15)

df[["Borrower_ID", "Industry", "Warning_Score"]]
```

```python
watchlist = df_clean[
    df_clean["Warning_Score"] >= 2
].sort_values(
    "Warning_Score",
    ascending=False
)

watchlist[
    ["Borrower_ID", "Industry", "Revenue_Growth_%",
     "Debt_to_EBITDA", "Interest_Coverage",
     "Current_Ratio", "Warning_Score"]
]
```
